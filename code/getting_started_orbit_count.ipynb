## Getting started on Orbit-Count

```python
import networkx as nx
import pandas as pd
import orbit_count
```

```python
G = nx.Graph()
G.add_edges_from(
    [
        ('v','1'),
        ('v','2'),
        ('v','3'),
        ('v','5'),
        ('v','6'),
        ('1','6'),
        ('4','5'),
        ('3','4')
    ]
)
```

```python
nodes_list = list(G.nodes())
```

```python
# Compute orbit count from n=2 to n=4
gdv_4 = orbit_count.node_orbit_counts(
    G,
    graphlet_size=4,
    node_list=nodes_list,
)

print("GDV shape:", gdv_4.shape)
```

```python
gdv_4_df = pd.DataFrame(
    gdv_4,
    index=nodes_list,
    columns=[f"orbit_{i}" for i in range(gdv_4.shape[1])]
)
gdv_4_df
```
