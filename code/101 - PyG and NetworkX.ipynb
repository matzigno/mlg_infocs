```python
mdg = nx.MultiDiGraph()
mdg.add_edge('A','B', relation_type = 'r1')
mdg.add_edge('A','B', relation_type = 'r2')
for edge in mdg.edges(data = True, keys = True):
    print(edge)
```

```python
mdg = nx.MultiDiGraph()
mdg.add_edge('A','B', key = 'r1')
mdg.add_edge('A','B', key = 'r2')
for edge in mdg.edges(data = True, keys = True):
    print(edge)
```
