## Naive Automorphis and Orbit Computation

```python
import networkx as nx
from itertools import permutations
```

```python
## Test Graph
G = nx.Graph()
G.add_edges_from([('u','v'),('v','w'),('u','w'),('w','z'),('z','u')])
```

```python
def get_automorphisms(G):
    nodes = tuple(G.nodes)
    permutes = list(permutations(nodes))
    automorph = []
    for pi in permutes:
        mapping = dict(zip(nodes, pi))
        for i,j in G.edges():
            if not G.has_edge(mapping[i],mapping[j]):
                break
        else:
            automorph.append(pi)
    return nodes, automorph             
```

```python
def get_orbits(nodes, automorphisms):
    orbits = set()
    for node in nodes:
        orbit_node = []
        for auto in automorphisms:
            mapping = dict(zip(nodes, auto))
            orbit_node.append(mapping[node])
        orbits.add(frozenset(orbit_node))
    return orbits
```

```python
get_orbits(*get_automorphisms(G))
```
